# Topic 1: Variables and Data Types

### Generation Prompt

> QuickRide charges a base fare of `$3.00, a per-mile rate of $1.50`, and a per-minute rate of $0.25. A customer just took a ride that was 12 miles and lasted 25 minutes. Can you write a Python program that stores all of these values and prints each one with a label? i have put in colab markdown but a per-mile rate looks off

```python
# Storing ride details as variables

base_fare = 3.00
per_mile_rate = 1.50
per_minute_rate = 0.25
distance = 12
time_minutes = 25

print("Base Fare:", base_fare)
print("Per-Mile Rate:", per_mile_rate)
print("Per-Minute Rate:", per_minute_rate)
print("Distance (miles):", distance)
print("Time (minutes):", time_minutes)
```

### Explanation

The code above creates five **variables** – named containers that each hold a single value:

1. `base_fare`, `per_mile_rate`, and `per_minute_rate` store the pricing rules. These are **floats** (decimal numbers like 3.00 and 1.50).
2. `distance` and `time_minutes` store the ride details. These are **integers** (whole numbers like 12 and 25).
3. Each `print()` statement displays a label followed by the value stored in that variable.

**Key takeaway:** Instead of scattering raw numbers throughout the code, we give each value a meaningful name. Anyone at QuickRide can read `per_mile_rate` and immediately know what it represents.

### Explanation Prompt
> Can you explain what variables and data types are? Why did we use names like `base_fare` instead of just writing the numbers directly?

```python
# Demonstrating why variables matter

# Without variables: hard to read and easy to make mistakes
print("Fare:", 3.00 + (1.50 * 12) + (0.25 * 25))

# With variables: clear and easy to update
base_fare = 3.00
per_mile_rate = 1.50
per_minute_rate = 0.25
distance = 12
time_minutes = 25

print("Fare:", base_fare + (per_mile_rate * distance) + (per_minute_rate * time_minutes))

print("\nBoth produce the same result, but the second version is readable and maintainable.")
```

### Explanation

A **variable** is a label attached to a value stored in memory. Think of it like a labeled jar in a kitchen – the jar marked "Sugar" tells you what is inside without opening it.

Python has several **data types**. The three we will use in this project are:

| Data Type | What It Stores | Example |
| --- | --- | --- |
| `int` | Whole numbers | `25` (minutes), `12` (miles) |
| `float` | Decimal numbers | `3.00` (dollars), `1.50` (rate) |
| `str` | Text | `"Aarav"` (name), `"completed"` (status) |

Using named variables instead of raw numbers makes code easier to read, update, and debug.

# Topic 2: Input, Output, and Formatting Essentials

### Generation Prompt
> Instead of hard-coding the ride details, I want the program to ask the driver to type in the customer's name, the distance in miles, and the time in minutes. Then it should display all the entered details back in a clean, readable format, like: "Ride Summary for Aarav: Distance = 12.5 mi, Time = 25 min."

```python
# Collecting ride details from user input

customer_name = input("Enter customer name: ")
distance = float(input("Enter distance in miles: "))
time_minutes = int(input("Enter time in minutes: "))

print(f"\nRide Summary for {customer_name}: Distance = {distance} mi, Time = {time_minutes} min.")
```

### Explanation

The code uses `input()` to ask the user for information instead of using fixed values:

1. **`input()`** displays a prompt and waits for the user to type something. It always returns a **string**.
2. **`float()`** converts the distance input from text to a decimal number (e.g., `"12.5"` becomes `12.5`).
3. **`int()`** converts the time input from text to a whole number (e.g., `"25"` becomes `25`).
4. The **f-string** `f"..."` lets us embed variables directly inside a string using `{}` curly braces.

Without the `float()` and `int()` conversions, Python would treat the inputs as text and any math on them would fail.

### Explanation Prompt
> Can you explain how the program takes input from the user and why we had to convert some inputs into numbers? Also, what is an f-string and how does it make printing easier?

```python
# Demonstrating why conversion matters

raw_input = "25"
print(f"Type of raw input: {type(raw_input)}")  # It is a string!

# This would NOT work for math:
# total = raw_input + 10  # TypeError!

# After conversion, math works:
converted = int(raw_input)
print(f"Type after int(): {type(converted)}")
print(f"Math works: {converted + 10}")

# f-string vs old way
name = "Aarav"
dist = 12.5
print("\nOld way: " + name + " traveled " + str(dist) + " miles.")  # Clunky
print(f"f-string: {name} traveled {dist} miles.")  # Clean and readable
```

### Explanation

`input()` always returns a string. Even if the user types `25`, Python sees it as the text `"25"`, not the number 25. That is why we wrap it with `int()` or `float()` before doing any math.

An **f-string** (formatted string literal) starts with `f` before the opening quote. Variables inside `{}` are automatically replaced with their values. This is much cleaner than gluing strings together with `+`.

### Modification Prompt
> Can you update the program so the output also shows the distance rounded to one decimal place and displays the time in a format like "025 min" (padded with zeros to three digits)? Also, print a separator line of dashes before and after the summary.

```python
# Formatted ride summary with separator lines

customer_name = "Aarav"
distance = 12.5
time_minutes = 25

print("-" * 50)
print(f"Ride Summary for {customer_name}")
print(f"  Distance : {distance:.1f} mi")
print(f"  Time     : {time_minutes:03d} min")
print("-" * 50)
```

### Explanation

Inside f-strings, format specifiers after a colon (`:`) control how values are displayed:

| Specifier | Meaning | Example |
| --- | --- | --- |
| `:.1f` | Round float to 1 decimal place | `12.5` |
| `:.2f` | Round float to 2 decimal places | `12.50` |
| `:03d` | Pad integer with zeros to 3 digits | `025` |
| `:05d` | Pad integer with zeros to 5 digits | `00025` |

The separator line `"-" * 50` repeats the dash character 50 times, creating a clean visual divider. These formatting techniques are essential for generating professional receipts and reports.

# Topic 3: Operators in Python

### Generation Prompt
> Now I want the program to actually calculate the total fare. Using the values we already have – base fare of $3.00, per-mile rate of $1.50, and per-minute rate of $0.25 – can you compute the fare for Aarav's ride of 12.5 miles and 25 minutes? Show the breakdown: the distance charge, the time charge, and the total fare. Also, if QuickRide offers a 10% discount, show the discounted fare.

```python
# Fare calculation with breakdown and discount

base_fare = 3.00
per_mile_rate = 1.50
per_minute_rate = 0.25
distance = 12.5
time_minutes = 25

# Calculate each component
distance_charge = per_mile_rate * distance
time_charge = per_minute_rate * time_minutes
total_fare = base_fare + distance_charge + time_charge

# Apply 10% discount
discount = total_fare * 0.10
discounted_fare = total_fare - discount

print(f"Distance Charge : ${distance_charge:.2f}")
print(f"Time Charge     : ${time_charge:.2f}")
print(f"Base Fare       : ${base_fare:.2f}")
print(f"Total Fare      : ${total_fare:.2f}")
print(f"Discount (10%)  : -${discount:.2f}")
print(f"Discounted Fare : ${discounted_fare:.2f}")
```

### Explanation

The code uses **arithmetic operators** to compute the fare step by step:

1. **`*` (multiplication):** `per_mile_rate * distance` gives the distance charge; `per_minute_rate * time_minutes` gives the time charge.
2. **`+` (addition):** `base_fare + distance_charge + time_charge` adds up all components.
3. **`*` and `-` together:** `total_fare * 0.10` calculates 10% of the total, and subtracting it gives the discounted fare.

Notice that all the variables from Topic 1 are being used together in calculations. That is why storing them with clear names matters.

### Explanation Prompt
> Can you explain the different operators used here? Are there other operators in Python that might be useful for fare calculations?

```python
# Demonstrating all useful operators for QuickRide

fare = 27.00

# Arithmetic operators
print("=== Arithmetic Operators ===")
print(f"Addition:       10 + 3 = {10 + 3}")
print(f"Subtraction:    10 - 3 = {10 - 3}")
print(f"Multiplication: 10 * 3 = {10 * 3}")
print(f"Division:       10 / 3 = {10 / 3:.4f}")
print(f"Integer Div:    10 // 3 = {10 // 3}")
print(f"Modulus:        10 % 3 = {10 % 3}")
print(f"Power:          2 ** 3 = {2 ** 3}")

# Comparison operators (return True or False)
print("\n=== Comparison Operators ===")
print(f"Is fare above $25?  {fare > 25}")
print(f"Is fare exactly $27? {fare == 27}")
print(f"Is fare below $20?  {fare < 20}")

# Assignment shortcut
print("\n=== Assignment Operator ===")
tip = 5.00
fare += tip  # Same as: fare = fare + tip
print(f"Fare after $5 tip: ${fare:.2f}")
```

### Explanation

Python provides several types of operators:

| Operator | Name | QuickRide Use Case |
| --- | --- | --- |
| `+` `-` `*` `/` | Basic math | Fare calculation |
| `//` | Integer division | Split fare into whole dollars per passenger |
| `%` | Modulus (remainder) | Find leftover cents after splitting |
| `**` | Power | Compound interest or growth calculations |
| `>` `<` `>=` `<=` `==` `!=` | Comparison | Check if fare exceeds a threshold |
| `+=` `-=` `*=` | Assignment shortcuts | Add tip to fare |

Comparison operators return `True` or `False` – these will become critical in the next topic when we add surge pricing.

### Modification Prompt
> The ride is shared between 3 passengers. Can you split the discounted fare equally and show: how much each person pays (rounded to 2 decimal places), and if there's any leftover amount (in cents) that can't be split evenly? Use both regular division and integer division to show the difference.

```python
# Splitting the fare among 3 passengers

discounted_fare = 24.38  # From previous calculation
passengers = 3

# Regular division (exact)
exact_split = discounted_fare / passengers
print(f"Exact split (/)  : ${exact_split:.6f}")
print(f"Rounded to 2 dec : ${round(exact_split, 2)}")

# Integer division (whole dollars only)
whole_dollars = discounted_fare // passengers
print(f"\nWhole dollars (//) : ${whole_dollars:.2f}")

# Modulus (leftover)
leftover = discounted_fare % passengers
leftover_cents = round(leftover * 100)
print(f"Leftover (%)       : ${leftover:.2f} ({leftover_cents} cents)")

print(f"\nEach passenger pays ${round(exact_split, 2)}, with {leftover_cents} cents remaining.")
```

### Explanation

This demonstrates the practical difference between division operators:

1. **`/` (regular division)** gives the exact decimal result. Useful when precision matters.
2. **`//` (integer division)** drops everything after the decimal. Shows how many whole dollars each person owes.
3. **`%` (modulus)** gives the remainder that cannot be split evenly.
4. **`round(value, 2)`** cleans up the display – customers do not want to see `$8.126667` on a receipt.

Fare splitting is a real feature in ride-sharing apps, and this is exactly how the math works behind the scenes.

# Topic 4: Conditional Statements in Python

### Generation Prompt
> QuickRide applies surge pricing during peak hours. If the ride happens between 8 AM to 10 AM or 5 PM to 9 PM, the total fare gets a 1.5x multiplier. If it's between 10 PM and 6 AM (late night), the multiplier is 1.25x. At all other times, there's no surge. Can you take the ride hour as input and compute the final fare using these rules? Use the same base fare, rates, and Aarav's ride details from before.

```python
# Fare calculation with surge pricing based on ride hour

base_fare = 3.00
per_mile_rate = 1.50
per_minute_rate = 0.25
distance = 12.5
time_minutes = 25

ride_hour = int(input("Enter the ride hour (0-23, e.g., 8 for 8 AM, 17 for 5 PM): "))

# Determine surge multiplier
if 8 <= ride_hour <= 9:
    surge_multiplier = 1.5
    period = "Morning Peak"
elif 17 <= ride_hour <= 20:
    surge_multiplier = 1.5
    period = "Evening Peak"
elif ride_hour >= 22 or ride_hour <= 5:
    surge_multiplier = 1.25
    period = "Late Night"
else:
    surge_multiplier = 1.0
    period = "Normal"

# Calculate fare
distance_charge = per_mile_rate * distance
time_charge = per_minute_rate * time_minutes
base_total = base_fare + distance_charge + time_charge
final_fare = base_total * surge_multiplier

print(f"\nRide Hour       : {ride_hour}:00 ({period})")
print(f"Base Total      : ${base_total:.2f}")
print(f"Surge Multiplier: {surge_multiplier}x")
print(f"Final Fare      : ${final_fare:.2f}")
```

### Explanation

The code uses **`if`, `elif`, and `else`** to make decisions based on the ride hour:

1. **`if 8 <= ride_hour <= 9`** checks for morning peak. If true, the surge multiplier is set to 1.5x.
2. **`elif 17 <= ride_hour <= 20`** checks for evening peak. This only runs if the first condition was false.
3. **`elif ride_hour >= 22 or ride_hour <= 5`** checks for late night. The `or` operator means either condition being true is enough.
4. **`else`** catches everything that did not match above – normal hours with no surge.

Python checks conditions **top to bottom** and stops at the first match. Think of it like a toll booth with lanes labeled "Morning Peak," "Evening Peak," "Late Night," and "Regular." Each ride goes through exactly one lane.

The fare formula from Topic 3 is reused here. Only the multiplier changes based on the time – this is the power of variables.

### Explanation Prompt
> Can you explain how if-elif-else works step by step? What happens if more than one condition could be true – which one does Python pick?

```python
# Tracing how Python evaluates conditions

ride_hour = 17  # 5 PM

print(f"Testing ride_hour = {ride_hour}\n")

print(f"Check 1: Is 8 <= {ride_hour} <= 9?  --> {8 <= ride_hour <= 9}")
print(f"  Result: SKIP (False)\n")

print(f"Check 2: Is 17 <= {ride_hour} <= 20? --> {17 <= ride_hour <= 20}")
print(f"  Result: MATCH! Surge = 1.5x (Evening Peak)")
print(f"  Python STOPS here. It does NOT check the remaining conditions.\n")

print(f"Check 3: (skipped) Is {ride_hour} >= 22 or {ride_hour} <= 5?")
print(f"Check 4: (skipped) else")

print("\nKey rule: Python picks the FIRST true condition and ignores the rest.")
```

### Explanation

When Python encounters an `if-elif-else` chain:

1. It evaluates each condition **from top to bottom**.
2. The moment it finds a condition that is `True`, it runs that block and **skips everything below**.
3. `else` is the catch-all – it only runs if every condition above was `False`. It does not need its own condition.

The **order matters**. If you placed a broader condition (like `else`) first, or put late-night before evening peak, the logic could produce wrong results.

### Modification Prompt
> QuickRide now also wants to give a loyalty discount. After applying surge pricing, if the customer has completed more than 20 rides, they get a 15% discount; if between 10 and 20 rides, they get a 10% discount; otherwise, no discount. Can you ask for the number of completed rides as input and apply this on top of the surge-adjusted fare? Print a final bill showing: base fare, distance charge, time charge, surge multiplier, surge-adjusted fare, loyalty discount percentage, discount amount, and final payable amount.

```python
# Complete fare calculation: surge pricing + loyalty discount

base_fare = 3.00
per_mile_rate = 1.50
per_minute_rate = 0.25
distance = 12.5
time_minutes = 25

ride_hour = int(input("Enter the ride hour (0-23): "))
completed_rides = int(input("Enter number of completed rides: "))

# --- Surge Pricing (Decision Block 1) ---
if 8 <= ride_hour <= 9:
    surge_multiplier = 1.5
    period = "Morning Peak"
elif 17 <= ride_hour <= 20:
    surge_multiplier = 1.5
    period = "Evening Peak"
elif ride_hour >= 22 or ride_hour <= 5:
    surge_multiplier = 1.25
    period = "Late Night"
else:
    surge_multiplier = 1.0
    period = "Normal"

# --- Loyalty Discount (Decision Block 2) ---
if completed_rides > 20:
    discount_pct = 15
elif completed_rides >= 10:
    discount_pct = 10
else:
    discount_pct = 0

# --- Fare Calculation ---
distance_charge = per_mile_rate * distance
time_charge = per_minute_rate * time_minutes
base_total = base_fare + distance_charge + time_charge
surge_fare = base_total * surge_multiplier
discount_amount = surge_fare * (discount_pct / 100)
final_fare = surge_fare - discount_amount

# --- Final Bill ---
print("\n" + "=" * 40)
print("       QUICKRIDE - FARE RECEIPT")
print("=" * 40)
print(f"  Base Fare         : ${base_fare:.2f}")
print(f"  Distance Charge   : ${distance_charge:.2f}")
print(f"  Time Charge       : ${time_charge:.2f}")
print(f"  Base Total        : ${base_total:.2f}")
print("-" * 40)
print(f"  Ride Hour         : {ride_hour}:00 ({period})")
print(f"  Surge Multiplier  : {surge_multiplier}x")
print(f"  Surge-Adj. Fare   : ${surge_fare:.2f}")
print("-" * 40)
print(f"  Completed Rides   : {completed_rides}")
print(f"  Loyalty Discount  : {discount_pct}%")
print(f"  Discount Amount   : -${discount_amount:.2f}")
print("=" * 40)
print(f"  FINAL PAYABLE     : ${final_fare:.2f}")
print("=" * 40)
```

### Explanation

The program now has **two independent decision blocks** applied in sequence:

1. **Surge Pricing:** Determines the multiplier based on the time of day.
2. **Loyalty Discount:** Determines the discount percentage based on ride history.

The layered logic works like an assembly line: first compute the base fare, then apply surge, then apply the loyalty discount on top. The order matters – discounting before surging would give a different (incorrect) result.

This final bill ties together every concept covered so far:
- **Variables (Topic 1)** store all the values.
- **Formatted output (Topic 2)** presents the receipt cleanly.
- **Operators (Topic 3)** do all the math.
- **Conditionals (Topic 4)** apply the right business rules.

### Validation Prompt
> Before calculating the fare, I want the program to check that: the distance is a positive number, the time is a positive number, the ride hour is between 0 and 23, and the number of completed rides is not negative. If any input is invalid, the program should show a clear error message and not proceed with the calculation.

```python
# Fare calculation with input validation

base_fare = 3.00
per_mile_rate = 1.50
per_minute_rate = 0.25

distance = float(input("Enter distance in miles: "))
time_minutes = int(input("Enter time in minutes: "))
ride_hour = int(input("Enter ride hour (0-23): "))
completed_rides = int(input("Enter number of completed rides: "))

# --- Input Validation ---
is_valid = True

if distance <= 0:
    print(f"Error: Distance ({distance}) must be a positive number.")
    is_valid = False

if time_minutes <= 0:
    print(f"Error: Time ({time_minutes}) must be a positive number.")
    is_valid = False

if ride_hour < 0 or ride_hour > 23:
    print(f"Error: Ride hour ({ride_hour}) must be between 0 and 23.")
    is_valid = False

if completed_rides < 0:
    print(f"Error: Completed rides ({completed_rides}) cannot be negative.")
    is_valid = False

if is_valid:
    # --- Surge Pricing ---
    if 8 <= ride_hour <= 9:
        surge_multiplier = 1.5
        period = "Morning Peak"
    elif 17 <= ride_hour <= 20:
        surge_multiplier = 1.5
        period = "Evening Peak"
    elif ride_hour >= 22 or ride_hour <= 5:
        surge_multiplier = 1.25
        period = "Late Night"
    else:
        surge_multiplier = 1.0
        period = "Normal"

    # --- Loyalty Discount ---
    if completed_rides > 20:
        discount_pct = 15
    elif completed_rides >= 10:
        discount_pct = 10
    else:
        discount_pct = 0

    # --- Calculate ---
    distance_charge = per_mile_rate * distance
    time_charge = per_minute_rate * time_minutes
    base_total = base_fare + distance_charge + time_charge
    surge_fare = base_total * surge_multiplier
    discount_amount = surge_fare * (discount_pct / 100)
    final_fare = surge_fare - discount_amount

    # --- Receipt ---
    print("\n" + "=" * 40)
    print("       QUICKRIDE - FARE RECEIPT")
    print("=" * 40)
    print(f"  Distance Charge   : ${distance_charge:.2f}")
    print(f"  Time Charge       : ${time_charge:.2f}")
    print(f"  Base Fare         : ${base_fare:.2f}")
    print(f"  Surge ({period:13s}): {surge_multiplier}x")
    print(f"  Loyalty Discount  : {discount_pct}% (-${discount_amount:.2f})")
    print("=" * 40)
    print(f"  FINAL PAYABLE     : ${final_fare:.2f}")
    print("=" * 40)
else:
    print("\nFare was NOT calculated due to invalid input.")
```

### Explanation

The code now validates every input before doing any calculation – just like how the QuickRide app would reject a ride request with a distance of -5 miles:

1. **Distance** must be positive. A ride of 0 or negative miles makes no sense.
2. **Time** must be positive. A ride cannot last 0 or negative minutes.
3. **Ride hour** must be between 0 and 23. There is no 25th hour in a day.
4. **Completed rides** cannot be negative. A customer cannot have taken -3 rides.

A flag variable `is_valid` tracks whether all checks passed. The fare is only calculated when every condition is satisfied. This is the "fail fast" principle: it is better to stop and show an error immediately than to compute a wrong fare and charge the customer.

This mirrors how professional software works: **validate first, calculate second, display last.**
